## FEATURE ENGINEERING

```python
#defining the group on min and max rule
def get_group(Age):
    if Age <= 12:
        return 'Child'
    elif Age <= 19:
        return 'Teen'
    elif Age <= 35:
        return 'Young Adult'
    elif Age <= 55:
        return 'Middle Aged'
    else:
        return 'Senior'

#appling it to the age colum
df_clean['Age_group']=df_clean['Age'].apply(get_group)
```

```python
print(df_clean[['Age','Age_group']].head(10).to_string(index=False))
```

```python
df_clean['Age_group'].unique()
```

```python
#customers by age grouyp
customer_by_age_group = df_clean.groupby('Age_group')['Customer_ID'].count()
customer_by_age_group

plt.figure(figsize=(5,4))

ax = customer_by_age_group.plot(kind='bar')

ax.bar_label(ax.containers[0],padding=2)

plt.tight_layout()
plt.show()
```

Middle Aged adults clearly dominate your customer base with a massive lead of 4,058 customers, followed by Young Adults (2,468) and Seniors (2,143). In contrast, the younger demographics are barely represented, with only 313 Teens and a tiny handful of 7 Children in the entire dataset.

```python
#customer by gender distribution
customer_by_Gender = df_clean.groupby('Gender')['Customer_ID'].count()
customer_by_Gender

plt.figure(figsize=(5,4))

ax = customer_by_Gender.plot(kind='bar')

ax.bar_label(ax.containers[0],padding=2)

plt.tight_layout()
plt.show()
```

Male customers form the larger segment of your customer base, accounting for 4,958 entries in the dataset. Female customers represent the smaller segment, totaling 4,031 entries, showing a noticeable gender gap in user acquisition.

```python
#customer by City distribution
customer_by_City = df_clean.groupby('City')['Customer_ID'].count()
customer_by_City

plt.figure(figsize=(5,4))

ax = customer_by_City.plot(kind='bar')

ax.bar_label(ax.containers[0],padding=2)

plt.tight_layout()
plt.show()
```

Kolkata completely dominates your customer market with a commanding lead of 2,255 customers. The remaining five cities (Bangalore, Chennai, Delhi, Hyderabad, and Mumbai) show a highly uniform split, hovering tightly between 1,335 and 1,358 customers each.

```python
df_clean = df[df['Age'].notna()].copy() 
```

```python
#custoemr lifetime , number of days the customer was active
df_clean['Days_Active']= (df_clean['Last_Purchase_Date']-df_clean['Signup_Date']).dt.days
#from the colums
target_columns = ['Days_Active', 'Purchase_Amount', 'Feedback_Score']

# 3. Print the average values for these columns (without the index formatting)
print("--- AVERAGE CUSTOMER METRICS ---")
print(df_clean[target_columns].mean().round().to_string())
```

```python
# getting the list of singup years
signup_years = sorted(df_clean['Signup_Date'].dt.year.dropna().unique())

signup_years
```

```python
# getting the list of unique purchase years
purchase_years = sorted(df_clean['Last_Purchase_Date'].dt.year.unique())

purchase_years
```

```python
#customer loyalty
# 1. Define your time brackets (in days)
day_bins = [0, 90, 180, 365, 730, 9999]
bin_labels = ['0-3 Months', '3-6 Months', '6-12 Months', '1-2 Years', '2+ Years']

# 2. Put each customer into a time bracket
df_clean['Time_Range'] = pd.cut(df_clean['Days_Active'], bins=day_bins, labels=bin_labels)

# 3. Print the clean count table (without the index numbers)
print("--- CUSTOMER COUNT BY TIME ACTIVE ---")
print(df_clean['Time_Range'].value_counts().to_string())

# 4. Draw a simple bar chart
df_clean['Time_Range'].value_counts().plot(kind='bar', color='orchid')

# 5. Label your chart simply
plt.title('Customer Distribution')
plt.ylabel('Number of Customers')
plt.show()

```

The chart shows that your business has excellent long-term retention, with the vast majority of customers remaining active for 2+ years. Shorter tenure brackets show progressively fewer customers, indicating that user churn significantly drops once a customer stays past their first year.
